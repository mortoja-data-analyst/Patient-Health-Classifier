```python
# ===================================================
# PROJECT 11: AI PATIENT HEALTH SEVERITY CLASSIFIER
# AUTHOR: MORTOJA SARDAR (DATA ANALYST)
# ===================================================

!pip install gTTS -q
import time, IPython.display as ipd
from gtts import gTTS

print("=" * 60)
print("     🏥 AI PATIENT HEALTH SEVERITY AUDITOR SYSTEM 🏥     ")
print("=" * 60 + "\n")

patient_name = "David Miller"
body_temperature_f = 103.5 # রোগীর শরীরের তাপমাত্রা (জ্বর)
heart_rate_bpm = 115       # হার্ট রেট

print(f"📥 [VITALS INGESTION] Patient Name: {patient_name}")
print(f"📥 [VITALS INGESTION] Temp: {body_temperature_f} F | Heart Rate: {heart_rate_bpm} BPM")
print("📊 [SYSTEM STATUS] Computing clinical triage severity metrics...")
time.sleep(2)

if body_temperature_f >= 102.0 or heart_rate_bpm >= 100:
    triage_severity = "CRITICAL EMERGENCY CONDITION (Immediate Doctor Required) 🚨"
    voice_msg = f"Emergency Alert Mortoja! Patient {patient_name} has critical high vital signs. Temperature is {body_temperature_f} and heart rate is {heart_rate_bpm}. Please notify the medical response team immediately."
else:
    triage_severity = "STABLE PATIENT VITALS COMPLIANT ✅"
    voice_msg = f"Mortoja, patient {patient_name} is stable."

print("-" * 60)
print(f"🔍 MEDICAL TARGET :  PATIENT EMERGENCY STATUS")
print("-" * 60)
print(f"📈 DIAGNOSED RISK :  {triage_severity}")
print("-" * 60 + "\n")

tts = gTTS(voice_msg, lang='en')
tts.save("patient_health_report.mp3")
ipd.Audio("patient_health_report.mp3")
```
